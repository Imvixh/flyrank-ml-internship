### Method choice

My question is whether observable content and search-performance features can identify pages that are likely to be declining.

I will compare three supervised classification methods:

1. **Logistic Regression** — a simple and interpretable learned baseline.
2. **Decision Tree** — captures nonlinear relationships while remaining easy to inspect.
3. **Random Forest** — provides a stronger nonlinear comparison and allows feature-importance inspection.

I chose these methods because the target is binary: a page is either observed as declining (`trend_direction = down`) or not declining.

The models will output a probability of decline rather than only a yes/no prediction. This probability can be used to rank pages for review.

I will select the strongest model using **Precision@50**, because the practical goal is to identify the most useful pages for review rather than simply maximize overall classification accuracy.

### Split design

I will use a **grouped-by-client split** so that content from the same client does not appear in both training and test sets.

This is an honest evaluation because pages belonging to the same client can share characteristics that would otherwise make the model appear stronger than it really is.

I will use an approximately **80/20 client-level train/test split**.

The target is created from `trend_direction`, where `down` represents observed decline. This target is used only for supervised learning and evaluation.

The features used for prediction will come from observable content and search-performance information available before the evaluation outcome.

I will avoid using current-window outcome fields or fields derived directly from the target, because those would create leakage.

The same test split will be used when comparing the learned models against the Week-4 baseline so that the comparison is fair.

```python
# ML-08 — Section 2: Split design

import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.model_selection import GroupShuffleSplit

# Load the dataset
data_path = ROOT / "data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(data_path)

print("Rows:", len(df))
print("Columns:", len(df.columns))

# Create the binary target.
# This is the observed evaluation outcome and is NOT used as an input feature.
df["is_declining"] = (
    df["trend_direction"].astype(str).str.lower() == "down"
).astype(int)

# Use client_id as the grouping variable.
groups = df["client_id"].astype(str)

# 80/20 grouped split.
# A client can appear in only one side of the split.
splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(df, y=df["is_declining"], groups=groups)
)

train = df.iloc[train_idx].copy()
test = df.iloc[test_idx].copy()

# Verify that no client appears in both sets.
train_clients = set(train["client_id"])
test_clients = set(test["client_id"])
overlap = train_clients.intersection(test_clients)

print("\n===== ML-08 SPLIT CHECK =====")
print(f"Train rows: {len(train):,}")
print(f"Test rows: {len(test):,}")
print(f"Train clients: {train['client_id'].nunique():,}")
print(f"Test clients: {test['client_id'].nunique():,}")
print(f"Client overlap: {len(overlap)}")

print("\nTrain declining rate:",
      f"{train['is_declining'].mean():.2%}")

print("Test declining rate:",
      f"{test['is_declining'].mean():.2%}")

if len(overlap) == 0:
    print("\nRESULT: PASS — no client appears in both train and test.")
else:
    print("\nRESULT: FAIL — client leakage detected.")
```

### Train and compare against my Week-4 baseline

I will train three supervised classifiers: Logistic Regression, Decision Tree, and Random Forest.

The models will use observable content and pre-window search-performance features. I will exclude current-window performance fields and the observed outcome fields from the feature set to reduce leakage risk.

The target is `is_declining`, where `trend_direction = down`.

I will compare all models with the Week-4 hand-written baseline on the **same client-grouped test split**.

The main ranking metric is **Precision@50** because the practical use case is to prioritize a small number of pages for human review. I will also report ROC-AUC as a secondary diagnostic.

The model with the strongest Precision@50 will be treated as the preferred decision-support model, while keeping the interpretation cautious: the model identifies observed patterns associated with decline; it does not prove causality or explain Google's ranking algorithm.

```python
# ML-08 — Section 3: Train models and compare with Week-4 baseline

import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import roc_auc_score


# ---------------------------------------------------------
# 1. Prepare train/test data
# ---------------------------------------------------------

target = "is_declining"

# Features must be observable before the evaluation outcome.
# Current-window performance and outcome-derived fields are excluded.
numeric_features = [
    "search_volume",
    "competition",
    "cpc",
    "word_count",
    "char_count",
    "impressions_prev_30d",
    "clicks_prev_30d",
    "sessions_prev_30d",
    "content_age_days",
    "age_tier_order",
    "days_since_last_update",
]

categorical_features = [
    "competition_level",
    "content_type",
    "main_intent",
    "age_tier",
    "freshness_tier",
    "word_count_tier",
    "char_count_tier",
    "impression_tier",
    "position_tier",
]

feature_columns = numeric_features + categorical_features

X_train = train[feature_columns].copy()
X_test = test[feature_columns].copy()

y_train = train[target].copy()
y_test = test[target].copy()


# ---------------------------------------------------------
# 2. Preprocessing
# ---------------------------------------------------------

numeric_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
    ]
)

categorical_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )),
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_pipeline, numeric_features),
        ("categorical", categorical_pipeline, categorical_features),
    ]
)


# ---------------------------------------------------------
# 3. Define the three models
# ---------------------------------------------------------

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=6,
        min_samples_leaf=50,
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        max_depth=10,
        min_samples_leaf=20,
        random_state=42,
        n_jobs=-1
    ),
}


# ---------------------------------------------------------
# 4. Train and evaluate models
# ---------------------------------------------------------

results = []
model_predictions = {}

def precision_at_k(y_true, scores, k=50):
    order = np.argsort(scores)[::-1]
    top_k = np.asarray(y_true)[order[:k]]
    return top_k.mean()


for name, model in models.items():

    pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", model),
        ]
    )

    pipeline.fit(X_train, y_train)

    probabilities = pipeline.predict_proba(X_test)[:, 1]

    p50 = precision_at_k(y_test, probabilities, k=50)
    auc = roc_auc_score(y_test, probabilities)

    results.append({
        "model": name,
        "precision_at_50": p50,
        "roc_auc": auc,
    })

    model_predictions[name] = probabilities


# ---------------------------------------------------------
# 5. Recreate the Week-4 baseline on THIS SAME test split
# ---------------------------------------------------------

baseline = test.copy()

baseline["prior_ctr_pct"] = np.where(
    baseline["impressions_prev_30d"] > 0,
    baseline["clicks_prev_30d"]
    / baseline["impressions_prev_30d"] * 100,
    np.nan
)

baseline["stale"] = (
    baseline["days_since_last_update"] >= 104
)

baseline["visible"] = (
    baseline["impressions_prev_30d"] >= 300
)

baseline["weak_ctr"] = (
    (baseline["prior_ctr_pct"] < 1.0)
    & baseline["prior_ctr_pct"].notna()
)

baseline["score"] = (
    3 * baseline["stale"].astype(int)
    + 2 * baseline["visible"].astype(int)
    + 2 * baseline["weak_ctr"].astype(int)
)

baseline_scores = baseline["score"].to_numpy()

baseline_p50 = precision_at_k(
    baseline[target].to_numpy(),
    baseline_scores,
    k=50
)


# ---------------------------------------------------------
# 6. Add baseline to the comparison table
# ---------------------------------------------------------

comparison = pd.DataFrame(results)

comparison = pd.concat(
    [
        pd.DataFrame([{
            "model": "Week-4 Baseline",
            "precision_at_50": baseline_p50,
            "roc_auc": np.nan,
        }]),
        comparison,
    ],
    ignore_index=True
)

comparison["precision_at_50"] = (
    comparison["precision_at_50"] * 100
)

comparison["roc_auc"] = (
    comparison["roc_auc"] * 100
)

comparison = comparison.rename(
    columns={
        "precision_at_50": "Precision@50 (%)",
        "roc_auc": "ROC-AUC (%)",
    }
)


# ---------------------------------------------------------
# 7. Display results
# ---------------------------------------------------------

print("===== ML-08 MODEL VS BASELINE =====")
print(f"Training rows: {len(train):,}")
print(f"Test rows: {len(test):,}")
print(f"Features used: {len(feature_columns)}")

display(
    comparison.sort_values(
        "Precision@50 (%)",
        ascending=False
    ).reset_index(drop=True)
)


# ---------------------------------------------------------
# 8. Identify the best model
# ---------------------------------------------------------

model_only = comparison[
    comparison["model"] != "Week-4 Baseline"
]

best_model_row = model_only.loc[
    model_only["Precision@50 (%)"].idxmax()
]

best_model_name = best_model_row["model"]

print(
    f"\nBest model by Precision@50: "
    f"{best_model_name}"
)

print(
    f"Best model Precision@50: "
    f"{best_model_row['Precision@50 (%)']:.2f}%"
)

print(
    f"Week-4 baseline Precision@50: "
    f"{baseline_p50 * 100:.2f}%"
)
```

### Errors and interpretation

The Random Forest was the strongest model on the held-out test split, achieving **56.0% Precision@50**, compared with **42.0% for the Week-4 baseline**.

This is a **14 percentage-point improvement** in the top-50 review queue.

The Random Forest also achieved a ROC-AUC of **67.71%**, slightly above the Decision Tree at 67.43% and clearly above Logistic Regression at 55.24%.

The model is likely benefiting from nonlinear relationships between content freshness, search visibility, engagement, and content characteristics. However, a top-ranked prediction can still be wrong: some pages may be flagged even though their observed trend is not declining.

I will treat these predictions as **decision-support recommendations**, not as proof that a page will decline or that a particular feature causes decline.

The main error cases are false positives in the top-ranked queue. These can occur when a page has strong observable risk signals but does not subsequently show the observed `down` trend.

Feature importance should therefore be interpreted as evidence of what the model relies on for prediction, not as causal evidence.

Overall, the measured result supports using the Random Forest as a stronger baseline candidate for ranking review priorities on this split, while retaining human review and further validation before operational use.

```python
# ML-08 — Section 4: Error analysis and feature interpretation

from sklearn.inspection import permutation_importance
import matplotlib.pyplot as plt


# ---------------------------------------------------------
# 1. Refit the best model so we can inspect it
# ---------------------------------------------------------

best_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestClassifier(
                n_estimators=200,
                max_depth=10,
                min_samples_leaf=20,
                random_state=42,
                n_jobs=-1
            )
        ),
    ]
)

best_pipeline.fit(X_train, y_train)

test_probabilities = best_pipeline.predict_proba(X_test)[:, 1]

test_predictions = (
    test_probabilities >= 0.5
).astype(int)


# ---------------------------------------------------------
# 2. Error analysis
# ---------------------------------------------------------

error_analysis = test[
    [
        "content_id",
        "days_since_last_update",
        "impressions_prev_30d",
        "clicks_prev_30d",
        "sessions_prev_30d",
        "trend_direction",
    ]
].copy()

error_analysis["actual"] = y_test.to_numpy()
error_analysis["predicted"] = test_predictions
error_analysis["probability_declining"] = test_probabilities

error_analysis["error_type"] = np.select(
    [
        (error_analysis["actual"] == 1)
        & (error_analysis["predicted"] == 1),

        (error_analysis["actual"] == 0)
        & (error_analysis["predicted"] == 1),

        (error_analysis["actual"] == 1)
        & (error_analysis["predicted"] == 0),

        (error_analysis["actual"] == 0)
        & (error_analysis["predicted"] == 0),
    ],
    [
        "true_positive",
        "false_positive",
        "false_negative",
        "true_negative",
    ],
    default="unknown"
)


print("===== ERROR ANALYSIS =====")

display(
    error_analysis["error_type"]
    .value_counts()
    .rename_axis("error_type")
    .reset_index(name="rows")
)


# ---------------------------------------------------------
# 3. Show the highest-confidence false positives
# ---------------------------------------------------------

false_positives = error_analysis[
    error_analysis["error_type"] == "false_positive"
].sort_values(
    "probability_declining",
    ascending=False
)

print("\n===== TOP FALSE POSITIVES =====")

display(
    false_positives.head(10)
)


# ---------------------------------------------------------
# 4. Show the highest-confidence false negatives
# ---------------------------------------------------------

false_negatives = error_analysis[
    error_analysis["error_type"] == "false_negative"
].sort_values(
    "probability_declining",
    ascending=False
)

print("\n===== TOP FALSE NEGATIVES =====")

display(
    false_negatives.head(10)
)


# ---------------------------------------------------------
# 5. Permutation importance
# ---------------------------------------------------------

print("\n===== PERMUTATION IMPORTANCE =====")

perm = permutation_importance(
    best_pipeline,
    X_test,
    y_test,
    scoring="roc_auc",
    n_repeats=5,
    random_state=42,
    n_jobs=-1
)

importance = pd.DataFrame({
    "feature": feature_columns,
    "importance_mean": perm.importances_mean,
    "importance_std": perm.importances_std,
})

importance = importance.sort_values(
    "importance_mean",
    ascending=False
).reset_index(drop=True)

display(
    importance.head(10)
)


# ---------------------------------------------------------
# 6. Plot top feature importance
# ---------------------------------------------------------

top_features = importance.head(10).sort_values(
    "importance_mean"
)

plt.figure(figsize=(9, 6))

plt.barh(
    top_features["feature"],
    top_features["importance_mean"]
)

plt.xlabel("Mean decrease in ROC-AUC")
plt.ylabel("Feature")
plt.title("Random Forest — Top 10 Permutation Importances")

plt.tight_layout()
plt.show()


# ---------------------------------------------------------
# 7. Final interpretation
# ---------------------------------------------------------

print("\n===== INTERPRETATION =====")

print(
    "The Random Forest is the strongest model on the held-out split "
    "by Precision@50."
)

print(
    f"Precision@50: "
    f"{precision_at_k(y_test, test_probabilities, 50):.2%}"
)

print(
    f"ROC-AUC: "
    f"{roc_auc_score(y_test, test_probabilities):.2%}"
)

print(
    "\nPermutation importance is used as a model-interpretation tool. "
    "It indicates which features the model relies on most for ranking "
    "performance, but it does not establish causality."
)
```
