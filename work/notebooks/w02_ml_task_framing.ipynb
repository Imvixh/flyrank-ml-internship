## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

### Answer

A fixed rule would be too rigid because content decline is influenced by several observable signals at the same time. Pages can differ in search demand, impressions, clicks, CTR, position, engagement, content age, and trend behaviour.

A single threshold such as "refresh every page with a negative trend" would ignore these interactions and could produce many unnecessary recommendations. ML can combine multiple signals and assign a relative risk or priority score, allowing pages to be ranked instead of using one fixed rule.

The model is therefore used as decision support for prioritizing human review, not as proof that refreshing a page will improve its performance.
